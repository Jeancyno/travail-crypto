```python
alphabet_original = list(
    "ABCDEFGHIJKLMNOPQRSTUVWXYZ"
    "abcdefghijklmnopqrstuvwxyz"
    " .,!?'-()"
    "0123456789"
    "@+_/#§&%"
)

print("Nombre de caractères :", len(alphabet_original))
```

```python
import random

rng = random.Random(79)

alphabet = rng.sample(alphabet_original, len(alphabet_original))

print("Alphabet désordonné :")
print("".join(alphabet))
```

```python
print("Nombre de caractères :", len(alphabet))
print("Nombre de caractères uniques :", len(set(alphabet)))

print(
    "Tous les caractères sont présents :",
    set(alphabet) == set(alphabet_original)
)
```

```python
char_to_num = {
    char: i for i, char in enumerate(alphabet)
}

num_to_char = {
    i: char for i, char in enumerate(alphabet)
}

caractere = 'A'

numero = char_to_num[caractere]
caractere_retour = num_to_char[numero]

print("Caractère original :", caractere)
print("Numéro :", numero)
print("Caractère retrouvé :", caractere_retour)
```

```python
a = 5
b = 8
m = 79

x = char_to_num['A']

y = (a * x + b) % m

print("Caractère original :", 'A')
print("Valeur numérique x :", x)
print("Valeur chiffrée y :", y)
print("Caractère chiffré :", num_to_char[y])
```

```python
def affine_encrypt_char(char, a, b):
    x = char_to_num[char]
    y = (a * x + b) % 79
    return num_to_char[y]
```

```python
print(affine_encrypt_char('A', 5, 8))
print(affine_encrypt_char('B', 5, 8))
print(affine_encrypt_char('a', 5, 8))
print(affine_encrypt_char(' ', 5, 8))
print(affine_encrypt_char('5', 5, 8))
```

```python
def affine_encrypt(text, a, b):
    result = ""

    for char in text:
        result += affine_encrypt_char(char, a, b)

    return result
    
```

```python
texte = "Bonjour jeancy Bakulu le boss"

texte_chiffre = affine_encrypt(texte, 5, 8)

print("Texte original :", texte)
print("Texte chiffré  :", texte_chiffre)
```

test d'inverse modulaire 

```python
a_inv = 16

y = char_to_num['Q']

x = (a_inv * (y - 8)) % 79

print("Valeur chiffrée y :", y)
print("Valeur retrouvée x :", x)
print("Caractère original :", num_to_char[x])
```

```python
def affine_decrypt_char(char, a, b):
    a_inv = pow(a, -1, 79)
    
    y = char_to_num[char]
    x = (a_inv * (y - b)) % 79
    
    return num_to_char[x]

print(affine_decrypt_char('Q', 5, 8))
```

```python
def affine_decrypt(text, a, b):
    result = ""

    for char in text:
        result += affine_decrypt_char(char, a, b)

    return result
```

```python
texte_chiffre = "QPr#PwU8#_§rVb8Q§kwBw8B_8lP  "

texte_dechiffre = affine_decrypt(texte_chiffre, 5, 8)

print("Texte chiffré   :", texte_chiffre)
print("Texte déchiffré :", texte_dechiffre)
```

CFB (cipher feedbak)

Tester CFB sur AAA

```python
a = 5
b = 8
iv = 12

texte = "AAA"

# Conversion des caractères en valeurs numériques
message = [char_to_num[c] for c in texte]

print("Valeurs du message :", message)
```

```python
cipher_values = []

previous = iv

for m_i in message:
    # Étape 1 : appliquer Affine au précédent
    feedback = (a * previous + b) % 79
    
    # Étape 2 : ajouter le caractère du message
    c_i = (feedback + m_i) % 79
    
    # Sauvegarder le résultat
    cipher_values.append(c_i)
    
    # Feedback : le résultat devient le précédent
    previous = c_i

print("Valeurs chiffrées :", cipher_values)

# Transformer les nombres en caractères

cipher_text = ''.join(num_to_char[c] for c in cipher_values)

print("Texte chiffré :", cipher_text)
```

```python
def cfb_encrypt(text, a, b, iv):
    cipher_values = []
    previous = iv

    for char in text:
        m_i = char_to_num[char]

        feedback = (a * previous + b) % 79

        c_i = (feedback + m_i) % 79

        cipher_values.append(c_i)

        previous = c_i

    return ''.join(num_to_char[c] for c in cipher_values)

    texte = "AAA"

resultat = cfb_encrypt(texte, 5, 8, 12)

print("Texte original :", texte)
print("Texte chiffré  :", resultat)
```

```python
texte = "Bonjour jeancy Bakulu  le boss"

resultat = cfb_encrypt(texte, 5, 8, 12)

print("Texte original :", texte)
print("Texte chiffré  :", resultat)
```

```python
def cfb_decrypt_char(cipher_char, previous, a, b):

    c_i = char_to_num[cipher_char]

    feedback = (a * previous + b) % 79

    m_i = (c_i - feedback) % 79

    return num_to_char[m_i]
```

```python
print(cfb_decrypt_char('%', 12, 5, 8))
```

faisons le parcour de tout les texte chiffré

```python
def cfb_decrypt(text, a, b, iv):
    result = ""
    previous = iv

    for cipher_char in text:

        plain_char = cfb_decrypt_char(
            cipher_char,
            previous,
            a,
            b
        )

        result += plain_char

        previous = char_to_num[cipher_char]

    return result
```

Première programmation RSA

```python
def rsa_keygen(p, q, e):
    # Calcul de n
    n = p * q

    # Calcul de phi(n)
    phi = (p - 1) * (q - 1)

    # Calcul de l'inverse modulaire de e
    d = pow(e, -1, phi)

    # Clé publique et clé privée
    public_key = (e, n)
    private_key = (d, n)

    return public_key, private_key
```

```python
def rsa_encrypt(message, public_key):
    e, n = public_key
    return pow(message, e, n)
```

```python
message = 10

chiffre = rsa_encrypt(message, public_key)

print("Message original :", message)
print("Message chiffré   :", chiffre)
```

Créons maintenant le déchiffrement


```python
def rsa_decrypt(ciphertext, private_key):
    d, n = private_key
    return pow(ciphertext, d, n)
    
```

```python
dechiffre = rsa_decrypt(chiffre, private_key)

print("Chiffré   :", chiffre)
print("Déchiffré :", dechiffre)
```

RSA ENCRYPT PARAMETRE 

```python
def rsa_encrypt_parameters(a, b, iv, public_key):
    encrypted_a = rsa_encrypt(a, public_key)
    encrypted_b = rsa_encrypt(b, public_key)
    encrypted_iv = rsa_encrypt(iv, public_key)

    return encrypted_a, encrypted_b, encrypted_iv
```

```python
encrypted_parameters = rsa_encrypt_parameters(
    a, b, iv, public_key
)

print("Paramètres originaux :", (a, b, iv))
print("Paramètres chiffrés  :", encrypted_parameters)
```

fonction de déchiffrement des paramètres.

```python
def rsa_decrypt_parameters(encrypted_parameters, private_key):
    encrypted_a, encrypted_b, encrypted_iv = encrypted_parameters

    a = rsa_decrypt(encrypted_a, private_key)
    b = rsa_decrypt(encrypted_b, private_key)
    iv = rsa_decrypt(encrypted_iv, private_key)

    return a, b, iv
```

definition de la fonction hybrid

```python
def hybrid_encrypt(text, a, b, iv, public_key):

    # Protection des paramètres avec RSA
    encrypted_parameters = rsa_encrypt_parameters(
        a, b, iv, public_key
    )

    # Chiffrement du texte avec Affine + CFB
    ciphertext = cfb_encrypt(text, a, b, iv)

    return encrypted_parameters, ciphertext
```

Décrypter

```python
def hybrid_decrypt(encrypted_parameters, ciphertext, private_key):

    # Récupération des paramètres avec RSA
    a, b, iv = rsa_decrypt_parameters(
        encrypted_parameters,
        private_key
    )

    print("Paramètres récupérés :", a, b, iv)

    # Déchiffrement du texte avec Affine + CFB
    plaintext = cfb_decrypt(ciphertext, a, b, iv)

    return plaintext
```
